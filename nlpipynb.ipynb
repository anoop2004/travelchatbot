```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from geotext import GeoText

# sentences with actual sources and destinations
data = [
    {"text": "I am traveling from New York to Los Angeles", "actual_source": "New York", "actual_destination": "Los Angeles"},
    {"text": "Flight from Mumbai to Delhi", "actual_source": "Mumbai", "actual_destination": "Delhi"},
    {"text": "Train ride from Boston to Washington", "actual_source": "Boston", "actual_destination": "Washington"},
    {"text": "Going to Paris from London", "actual_source": "London", "actual_destination": "Paris"},
    {"text": "Driving from Chicago to Toronto", "actual_source": "Chicago", "actual_destination": "Toronto"},
]

def extract_src_dest(text):
    places = GeoText(text)

    if len(places.cities) < 2:
        return None, None

    place1, place2 = places.cities[:2]
    first_half, second_half = text.split(place1, 1)


    destinations = ["to", "arriving", "reaching", "destination"]
    for word in first_half.split():
        if word in destinations:
            return place2, place1
    return place1, place2

# Evaluate accuracy
correct = 0
for entry in data:
    predicted_source, predicted_destination = extract_src_dest(entry["text"])

    if predicted_source == entry["actual_source"] and predicted_destination == entry["actual_destination"]:
        correct += 1

accuracy = correct / len(data) * 100
print(f"Accuracy of Source-Destination Extraction: {accuracy:.2f}%")
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from gensim.models import Word2Vec
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
data = []
for category, questions in categories.items():
    for question in questions:
        data.append([category, question])
df = pd.DataFrame(data, columns=["target", "text"])

# Preprocessing function
def preprocess(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)  # Remove numbers
    text = text.translate(str.maketrans("", "", string.punctuation))  # Remove punctuation
    text = text.strip()
    return text

def stopword_removal(text):
    words = word_tokenize(text)
    return [word for word in words if word not in stopwords.words("english")]

df["preprocessed_text"] = df["text"].apply(lambda x: stopword_removal(preprocess(x)))

word2vec_model = Word2Vec(sentences=df["preprocessed_text"], vector_size=100, window=5, min_count=1, workers=4)

def vectorize_text(text):
    vectors = [word2vec_model.wv[word] for word in text if word in word2vec_model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(100)  # Handle empty cases

X = np.array([vectorize_text(words) for words in df["preprocessed_text"]])

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(df["target"])

X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

classifiers = {
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=500),
    "SVM": SVC(kernel='linear', probability=True),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, random_state=42)
}

for name, clf in classifiers.items():
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"{name} Accuracy: {accuracy:.4f}")

sample_query = input("Enter the text: ").strip()
if sample_query:
    sample_preprocessed = stopword_removal(preprocess(sample_query))
    sample_vector = vectorize_text(sample_preprocessed)

    best_classifier = max(classifiers.items(), key=lambda x: x[1].score(X_test, y_test))[1]  # Select best model
    sample_prediction = best_classifier.predict([sample_vector])
    predicted_category = label_encoder.inverse_transform(sample_prediction)
    print("Predicted Category:", predicted_category[0])
else:
    print("No input provided!")
```

```python
data = []
for category, questions in categories.items():
    for question in questions:
        data.append([category, question])

df = pd.DataFrame(data, columns=["target", "text"])

category_dict = {1: "flight", 2: "hostel", 3: "plan", 4: "activities", 5: "food"}

def preprocess(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)  # Remove digits
    text = text.translate(str.maketrans("", "", string.punctuation))  # Remove punctuation
    text = text.strip()
    return text

def stopword_removal(text):
    words = word_tokenize(text)
    return ' '.join([word for word in words if word not in stopwords.words('english')])

df["preprocessed_text"] = df["text"].apply(lambda x: stopword_removal(preprocess(x)))

tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(df["preprocessed_text"])

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(df["target"])

X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y_encoded, test_size=0.2, random_state=42)

classifiers = {
    "RandomForest": RandomForestClassifier(n_estimators=100, random_state=42),
    "LogisticRegression": LogisticRegression(max_iter=200),
    "SVM": SVC(kernel='linear', probability=True),
    "KNN": KNeighborsClassifier(n_neighbors=3),
    "GradientBoosting": GradientBoostingClassifier()
}

best_model = None
best_accuracy = 0

for name, clf in classifiers.items():
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"{name} Accuracy: {accuracy:.4f}")

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_model = clf

print(f"Best Model: {best_model.__class__.__name__} with Accuracy: {best_accuracy:.4f}")

sample_query = input("Enter the text: ").strip()
if sample_query:
    sample_preprocessed = stopword_removal(preprocess(sample_query))
    sample_tfidf = tfidf_vectorizer.transform([sample_preprocessed])
    sample_prediction = best_model.predict(sample_tfidf)
    predicted_category = label_encoder.inverse_transform(sample_prediction)
    print("Predicted Category:", category_dict[predicted_category[0]])
else:
    print("No input provided!")
```
